Importación de Librerías y Carga de Datos

```python
import pandas as pd
import numpy as np

# Cargar el archivo raw desde la ruta relativa
df = pd.read_csv('../data/candidates.csv', sep=';')
df.head()
```

Inspección General y Tipos de Datos

```python
print(f"Dimensiones del dataset: {df.shape[0]} filas, {df.shape[1]} columnas\n")
print("Estructura de tipos de datos:")
df.info()
```

Auditoría de Calidad (nulos y duplas)

```python
print("--- Conteo de Valores Nulos ---")
print(df.isnull().sum())

print("\n--- Conteo de Filas Duplicadas ---")
print(f"Total filas duplicadas: {df.duplicated().sum()}")
```

Estadísticas Descriptivas de Variables Numéricas

```python
# Análisis descriptivo de puntuaciones y experiencia
df[['Code Challenge Score', 'Technical Interview Score', 'YOE']].describe().T
```

Análisis de Variables Categóricas

```python
print(f"Cantidad de Tecnologías únicas: {df['Technology'].nunique()}")
print(f"Cantidad de Países únicos: {df['Country'].nunique()}")
print("\nValores en Seniority:")
print(df['Seniority'].value_counts())
```

Validación de Regla de Negocio (Is_Hired)

```python
# Aplicar regla de negocio: ambos puntajes >= 7
df['Is_Hired'] = ((df['Code Challenge Score'] >= 7) & (df['Technical Interview Score'] >= 7)).astype(int)

print("Distribución Global de Contratación:")
print(df['Is_Hired'].value_counts())

print("\nPorcentaje de Conversión:")
print((df['Is_Hired'].value_counts(normalize=True) * 100).round(2).astype(str) + '%')
```
